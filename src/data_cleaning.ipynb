```python
import pandas as pd
import numpy as np
import os
```

```python
DATA_DIR  = "../../../data/IDRiD"
train_dir = os.path.join(DATA_DIR, "Train", "images", "caption")
test_dir  = os.path.join(DATA_DIR, "Test", "images", "caption")
train_csv = os.path.join(DATA_DIR, "Train", "annotations.csv")
test_csv = os.path.join(DATA_DIR, "Test", "annotations.csv")

train_df = pd.read_csv(train_csv)
test_df  = pd.read_csv(test_csv)
```

```python
# Check for missing text in captions
missing_text = train_df["caption"].isna() | (train_df["caption"] == "")
missing_count = missing_text.sum()
total_count = len(train_df)
print(missing_count, total_count)
```

```python
# check for missing images

missing_imgaes = train_df["Image name"].isna()
missing_count = missing_imgaes.sum()
total_count = len(train_df)
print(missing_count, total_count)
```
